#### Login user program

```python
email = input("Enter the email:")
password = input("Enter the password:")

if email == 'bilal@gmail.com' and password == '1234':
    print('WELCOME')
else:
    print('User not found')
```

```python
email = input("Enter the email:")
password = input("Enter the password:")

if email == 'bilal@gmail.com' and password == '1234':
    print('WELCOME')
elif email == 'bilal@gmail.com' and password != '1234':
    print("Incorrect password")
    password = input("Enter password again:")
    if password != '1234':
        print("WELCOME")
    else:
        print("BETA TU NA KR PAE GA")
else:
    print('User not found')
```
