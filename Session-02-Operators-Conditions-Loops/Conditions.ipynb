### If else statment
when there is branching we will use if else statments

#### Login user program

```python
email = input("Enter the email:")
password = input("Enter the password:")

if email == 'bilal@gmail.com' and password == '1234':
    print('WELCOME')
else:
    print('User not found')
```

## Nested loop

loop within loop

```python
email = input("Enter the email:")
password = input("Enter the password:")

if email == 'bilal@gmail.com' and password == '1234':
    print('WELCOME')
elif email == 'bilal@gmail.com' and password != '1234':
    print("Incorrect password")
    password = input("Enter password again:")
    if password == '1234':
        print("WELCOME")
    else:
        print("BETA TU NA KR PAE GA")
else:
    print('User not found')
```

## Find the minimum of three numbers enter by user

```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))
num3 = int(input("Enter third number:"))

if num1 <= num2 and num1 <= num3:
    print("Number one is Smaller")
elif num2 <= num3:
    print("Number two is smaller")
else:
    print("three is small")
           
```

### Menu driven Calculator

```python
fnum = int(input("Enter the first number"))
snum = int(input("Enter the second number"))

op = input("enter the operation")

if op == '+':
    print(fnum+snum)
elif op == '-':
    print(fnum-snum)
elif op == '/':
    print(fnum/snum)
elif op == '*':
    print(fnum*snum)
else:
    print(fnum%snum)
```
