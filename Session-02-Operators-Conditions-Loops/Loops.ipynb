### While loop

A while loop is a computer code block that runs over and over again as long as a specific rule or condition stays true.

## Program to print a table of a given number by the user

```python
num = int(input("Enter the number"))

i = 1
while i<11:
    print(num, '*' , i , '=' , num*i)
    i+=1
    
```

### While loop with else statment

```python
x = 1
while x<3:
    print(x)
    x += 1
else:
    print("limit crossed")
```

## FOR LOOP
- A for loop is a programming tool used to repeat a block of code a specific number of times.
- A for loop in Python is used to iterate over a sequence (such as a list, tuple, dictionary, string, or range) and execute a block of code for each item in that sequence.

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(1,20,2):
    print(i)
```

```python
for item in [1,4,2,5,84,5,7]:
    print(item)
```

#### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
curr_pop = 10000

for i in range(10,0,-1):
    print(i,'=',curr_pop)
    curr_pop = curr_pop - 0.1*curr_pop
```
