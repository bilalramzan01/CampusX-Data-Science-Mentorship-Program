### While loop

A while loop is a computer code block that runs over and over again as long as a specific rule or condition stays true.

## Program to print a table of a given number by the user

```python
num = int(input("Enter the number"))

i = 1
while i<11:
    print(num, '*' , i , '=' , num*i)
    i+=1
    
```

### While loop with else statment

```python
x = 1
while x<3:
    print(x)
    x += 1
else:
    print("limit crossed")
```
