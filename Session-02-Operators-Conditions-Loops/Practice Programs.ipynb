**Problem 1:** Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

Salary(Lakhs) : Tax(%)

- Below 5 : 0%
- 5-10 : 10%
- 10-20 : 20%
- aboove 20 : 30%

```python
salary = int(input("Enter the salary"))

if salary < 5000000:
    print("Full salary, no tax:", salary)
elif salary >5000000 and salary <1000000:
    salary = salary*0.1
    print("10% tax on salary:" ,salary)
elif salary > 1000000 and salary < 2000000:
    salary = salary*0.2
    print("20% tax :" ,salary)
else:
    salary = salary*0.3
    print("30% Tax:", salary)
```
