**Problem 1:** Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

Salary(Lakhs) : Tax(%)

- Below 5 : 0%
- 5-10 : 10%
- 10-20 : 20%
- aboove 20 : 30%

```python
ctc = int(input("Enter the salary"))

hra = ctc * 10 / 100
da = ctc * 5 / 100
pf = ctc * 3 / 100

# Step 2: Determine tax rate based on which slab the CTC falls into

if ctc < 5:
    tax_rate = 0
elif ctc < 10:
    tax_rate = 10
elif ctc < 20:
    tax_rate = 20
else:
    tax_rate = 30

tax = ctc * tax_rate / 100

in_hand = ctc - hra - da - pf - tax
print(in_hand)
```

**Problem 2:** Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

```python
angle1 = int(input("Enter the angle 1:"))
angle2 = int(input("Enter the angle 2:"))
angle3 = int(input("Enter the angle 3:"))

if angle1 > 0 and angle2 > 0 and angle3 > 0 and (angle1 + angle2 + angle3 == 180):
    print("Valid triangle")
else:
    print("invalid traingle")
```

**Problem 5 - Exercise 12: Display Fibonacci series up to 10 terms.**

Note: The Fibonacci Sequence is a series of numbers. The next number is found by adding up the two numbers before it. The first two numbers are 0 and 1. For example, 0, 1, 1, 2, 3, 5, 8, 13, 21. The next number in this series above is 13+21 = 34

```python
a = 0
b = 1

print("Fibonacci series up to 10 terms:")
for i in range(10):
    print(a)
    next_term = a + b
    a = b
    b = next_term
```

**Problem 6 - Find the factorial of a given number.**
Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: !) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```python
num = int(input("Enter the number"))

fact = 1
for i in range( 1, num+1):
    fact = i * fact
print(fact)
    
```

**Problem 8:** Take a user input as integer N. Find out the sum from 1 to N. 
If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

```python
n = int(input("Enter the number: "))
sum = 0
i = 1
while i <= n:
    if sum > 300:
        break
    if i % 5 != 0:
        sum = sum + i
    i = i + 1
print(sum)
```
