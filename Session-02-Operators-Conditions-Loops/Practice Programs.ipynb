**Problem 1:** Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

Salary(Lakhs) : Tax(%)

- Below 5 : 0%
- 5-10 : 10%
- 10-20 : 20%
- aboove 20 : 30%

```python
ctc = int(input("Enter the salary"))

hra = ctc * 10 / 100
da = ctc * 5 / 100
pf = ctc * 3 / 100

# Step 2: Determine tax rate based on which slab the CTC falls into

if ctc < 5:
    tax_rate = 0
elif ctc < 10:
    tax_rate = 10
elif ctc < 20:
    tax_rate = 20
else:
    tax_rate = 30

tax = ctc * tax_rate / 100

in_hand = ctc - hra - da - pf - tax
print(in_hand)
```

**Problem 8:** Take a user input as integer N. Find out the sum from 1 to N. 
If any number if divisible by 5, then skip that number. And if the sum is greater than 300, don't need to calculate the sum further more. Print the final result. And don't use for loop to solve this problem.

```python
n = int(input("Enter the number: "))
sum = 0
i = 1
while i <= n:
    if sum > 300:
        break
    if i % 5 != 0:
        sum = sum + i
    i = i + 1
print(sum)
```

**Problem 13:Print all the Armstrong numbers in a given range.**
Range will be provided by the user
Armstrong number is a number that is equal to the sum of cubes of its digits. For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

```python
start = int(input("Enter starting range: "))
end = int(input("Enter ending range: "))

for num in range(start, end + 1):

    original = num
    sum = 0

    while num > 0:
        digit = num % 10          # get last digit
        sum = sum + digit ** 3    # add cube of digit
        num = num // 10           # remove last digit

    if sum == original:
        print(original, end=" ")
```

### Problem 14:Calculate the angle between the hour hand and minute hand.
Note: There can be two angles between hands; we need to print a minimum of two. Also, we need to print the floor of the final result angle. For example, if the final angle is 10.61, we need to print 10.

Input:
H = 9 , M = 0
Output:
90
Explanation:
The minimum angle between hour and minute hand when the time is 9 is 90 degress.

```python
H = int(input("Enter hour: "))
M = int(input("Enter minutes: "))

# Hour hand angle
hour_angle = (H * 30) + (M * 0.5)

# Minute hand angle
minute_angle = M * 6

# Difference between both hands
angle = abs(hour_angle - minute_angle)

# Minimum angle
if angle > 180:
    angle = 360 - angle

# Floor the answer
print(int(angle))
```
