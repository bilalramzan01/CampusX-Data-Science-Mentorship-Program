# Operators in python

the operation performed on more then one operands called operators



### Arthmetic Operators

```python
print(5+7)
print(5-7)
print(5*7)
print(15/7)
print(15//7)  #Convert into integer
print(5%7)

```

### Comparison/Relational Operators

```python
print(5>8) # Greater then
print(5<8) # smaller then
print(5<=8) #smaller and equal
print(5>=8) #gretaer and equal
# print(5=8) # assignment operator
print(5 == 5) # equal to
print(5!=8) # not equal to
```

### Logical operator



```python
print (1 and 0) # AND operator, output zero when both zero
print (1 or 0) # OR operator, output 1 when only one true
print( not 1) # reverse the input
```

### Bit Wwise operators

operators that works on binary values

```python
print(2 & 3) # 2 is 010 and 3 is 110.when we make them combine it give 010 which means it gives 2 answer.
print(2 | 3) #Bitwise OR.
print(2 ^ 3) #Bitwise xor
print(~3) #Bitwise not
print(2<<3) #bitwise left
print(2>>3) # Bitwise right
```

### Assignment operator

```python
a = 10
a
```

```python
a = 10
a += 2
a
```

### Membership operator

```python
# in and notin

print('D' in 'Delhi')
print('D' not in 'Delhi')
print(123 in [2,3,5,4332])
```
