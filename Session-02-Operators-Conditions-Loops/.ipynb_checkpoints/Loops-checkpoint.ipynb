### While loop

A while loop is a computer code block that runs over and over again as long as a specific rule or condition stays true.

## Program to print a table of a given number by the user

```python
num = int(input("Enter the number"))

i = 1
while i<11:
    print(num, '*' , i , '=' , num*i)
    i+=1
    
```

### While loop with else statment

```python
x = 1
while x<3:
    print(x)
    x += 1
else:
    print("limit crossed")
```

## Guessing Game

```python
import random
jackpot = random.randint(1,100)

guess = int(input("Guess the number"))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print("Oper aoo yar")
    else:
        print("Neche jao")

    guess = int(input("Guess the number"))
    counter += 1

else:
    print("Correct Guess")
    print("You find in attempts:", counter)
jackpot
```

## FOR LOOP
- A for loop is a programming tool used to repeat a block of code a specific number of times.
- A for loop in Python is used to iterate over a sequence (such as a list, tuple, dictionary, string, or range) and execute a block of code for each item in that sequence.

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(1,20,2):
    print(i)
```

```python
for item in [1,4,2,5,84,5,7]:
    print(item)
```

#### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
curr_pop = 10000

for i in range(10,0,-1):
    print(i,'=',curr_pop)
    curr_pop = curr_pop/1.1
```

## Nested Loops

### Sequence Sum

1/1!+2/2!+3/3!+....

```python
num = int(input("Enter the number"))
result = 0 
fact = 1
for i in range(1,num+1):
    fact = fact * i
    result = result + i/fact
print(result)
    
    
```

### All possible pairs

```python
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

### Print patterns

```python
num = int(input('Enter no of rows'))
for i in range(1,num+1):
    print(i*'*')
```

```python
num = int(input('Enter no of rows'))

for i in range(1,num+1):
    for j in range(1, i+1):
        print("*", end='')
    print()
```

```python
rows = int(input("Enter no of rows"))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print(j, end="")
    for k in range(i-1,0,-1):
        print(k, end='')
    print()
```

## Loop control statments


### 1. Break Statment

- Can break or stop the loop when condition meet
- 

```python
for i in range (1,10):
    if i == 5:
        break
    print(i)
    
    
```

### User will give you a range and find out prime numbers

```python
lower = int(input("Enter the lower range:"))
upper = int(input("Enter the upper range:"))

for i in range(lower,upper+1):
    for j in range(2,i):
        if i%j == 0:
            break
    else:
            print("Prime number:", i)
```

### Continue Statment

skip the true statment and complete the iterations

```python
for i in range(1,10):
    if i == 5:
        continue
    print(i)
```
