## Pandas DataFrame
- Tabular data in a pandas, having many columns is called Data Frames in Pandas.
- Table (Dataframes in pandas)
- every row and column is a series

```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

#### How to create a Data Frame
- It will be created using 2D list

```python
student_data = ([
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
])
```

```python
student_data
```

```python
pd.DataFrame(student_data, columns=['iq','marks','package'])
```

### Creating using dict

```python
student_dict = {
    'iq' : [100,90,120,80],
    'marks' : [80,70,100,50],
    'package' :[10,7,14,2]
}
```

```python
students = pd.DataFrame(student_dict)
```

```python
students
```
