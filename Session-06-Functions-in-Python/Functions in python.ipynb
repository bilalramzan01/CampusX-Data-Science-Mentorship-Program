## Functions in python

A function is a named block of reusable code that performs a specific task. Instead of writing the same lines of code again and again, you write them once inside a function, give it a name, and then just call that name whenever you need that task done.

Think of it like a recipe: you write the recipe once, and every time you want that dish, you just follow the recipe — you don't rewrite it from scratch.

```python
# Function creation
def iseven(num):
    ''' 
    This function returns if the given number is off or even
    '''
    if type(num) == int:
        if num%2==0:
            return 'even'
        else:
            return 'odd'
    else:
        return 'Only integers Allowed'
```

```python
#  Fucntion calling
for i in range(1,11):
    x = iseven(i)
    print(x)
```

### Types of arguments

#### Default arguments

```python
def power(a,b):
    return a**b
power(2,3)
```

```python
power(2)
```

```python
def power(a=1,b=1):
    return a**b
power(2)
```

#### Positional arguments

```python
# Arguments follow sequence and order. First value assign to first variable and 2nd value to 2nd variable.
def power(a,b):
    return a**b
power(2,3)
```

#### Keyword Arguments

```python
# Ignore the position
def power(a,b):
    return a**b
power(b=2,a=3)
```

```python
# arg: Allow us to pass a variable number of non-keyword arguments to a function
# vs
# kwargs: Allow us to pass any number of keyword arguments.
```

#### Args

```python
def multiply(a,b,c):
    return a*b*c
multiply(3,34,4)
```

```python
def multiply(*args):
    product = 1
    for i in args:
        product = product * i

    return product
multiply(3,34,4,3,2,4,5,3,5)
```

#### kwargs
kwargs: Allow us to pass any number of keyword arguments.

```python
def display(**kwargs):
    for (key,value) in kwargs.items():
        print(key,'->',value)

display(pakistan='Islamabad', India='delhi', USA='washington')
```
