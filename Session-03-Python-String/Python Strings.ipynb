## Python string

#### Creating python strings

```python
a = 'Hello' 
a = "Hello"  
a = '''Hello'''
```

```python
# Why need single and double both

a = " it's raining outside"

# as ' is in the sentence thats why we need both

# why triple line

a = ''' Hello
        My name is Bilal.
        and i am ai engineer '''

# for multi line string we use ''''
```

#### Accessing substring from a string



```python
# Using Positive indexing

a = 'hello world'
print(a[1])

# starting from 0 and going right to left
```

```python
# using negative indexing

a = 'hello world'
print(a[-1])

# starting from 1 and going left to right
```

### Slicing

```python
a = 'hello world'
print(a[0:5])
```

```python
a = 'hello world'
print(a[0:5:2])
```
