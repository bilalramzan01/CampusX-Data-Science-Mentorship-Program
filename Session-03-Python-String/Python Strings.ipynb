#### Accessing substring from a string



```python
# Using Positive indexing

a = 'hello world'
print(a[1])

# starting from 0 and going right to left
```

```python
# using negative indexing

a = 'hello world'
print(a[-1])

# starting from 1 and going left to right
```

### Slicing

```python
a = 'hello world'
print(a[0:5])
```

```python
a = 'hello world'
print(a[0:5:2])
```

```python
a = 'hello world'
print(a[-5:])
```

```python
a = "121"

b = a[::-1]

if a == b:
    print("Palindrome")
else:
    print("not palindrom")
```
