## Sets
- Holds unique values
- unordered
- mutuable
- no duplicates allowed
- can't contain mutuable data types
- creted by {} 

```python
# Empty sets
s = set()
```

```python
type(s)
```

```python
# 1D homogenius set
s = {1,2,3,4,5}
```

```python
type(s)
```

```python
# Hetrogenius set

s = {1,2,3,'Bilal'}
```

```python
s
```

```python
# Ignore true because true value is 1 and duplicates not allowed.
s = {2,6,9,3,'Bilal', True}
```

```python
s
```
