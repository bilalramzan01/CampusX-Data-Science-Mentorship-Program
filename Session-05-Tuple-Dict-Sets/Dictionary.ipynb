## Dictionary

- data type used to store key value pairs 
- Assoicated array
- Mutuable
- no indexing 
- keys can't be repeated.
- keys can't be mutuable

```python
# Empty dict
```

```python
d = {}
type(d)
```

```python
# D1

d1 = { 'Name':'Bilal','Age':23}
d1
```

```python
# 2D 

d2 = {
    'name' : 'Bilal',
    'college' : 'PGC' ,
    'Marks' : {'dsa':50,'OOP':80}
}
d2
```

```python
d4 = dict([('Name','Bilal'),(2,2)])
d4
```

```python
d = {
    'name' : 'Bilal',
    'college' : 'PGC' ,
    'Marks' : {'dsa':50,'OOP':80}
}
```

```python
# Accessing items by key
d['name']
```

```python
d.get('name')
```

#### Adding items

```python
d['City'] = 'Lahore'
d
```

#### Deleting items

```python
d.popitem()
d
```

```python
del d['name']
d
```

```python
d.clear()
```

```python
d
```

```python
d2 = {
    'name' : 'Bilal',
    'college' : 'PGC' ,
    'Marks' : {'dsa':50,'OOP':80}
}
d2
```

```python
# Accessing D2 items
d2['Marks']['dsa']
```

```python
d2['Marks']['math'] = 70
d2
```

```python
# Edit
d2['Marks']['math'] = 75
d2
```

```python
d2['Marks']['dsa'] = 98
d2
```

### Dict operations (working on keys)

```python
'Bilal' in d2
```

```python
'name' in d2
```

```python
for i in d2:
    print(i,d2[i])
```
