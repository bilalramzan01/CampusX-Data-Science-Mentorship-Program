## Dictionary

- data type used to store key value pairs 
- Assoicated array
- Mutuable
- no indexing 
- keys can't be repeated.
- keys can't be mutuable

```python
# Empty dict
```

```python
d = {}
type(d)
```

```python
# D1

d1 = { 'Name':'Bilal','Age':23}
d1
```

```python
# 2D 

d2 = {
    'name' : 'Bilal',
    'college' : 'PGC' ,
    'Marks' : {'dsa':50,'OOP':80}
}
d2
```

```python
d4 = dict([('Name','Bilal'),(2,2)])
d4
```
