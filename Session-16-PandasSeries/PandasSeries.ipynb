### Pandas Series
Pandas series is a row or column in a table. It is 1D array holding data of any type.

```python
import numpy as np
import pandas as pd
```

```python
# Create a list of array ( countries list)

country = ['Pakistan', 'India','Nepal','USA','Afghanistan','Italy']
```

```python
# Make series of countries
```

```python
pd.Series(country)
```

```python
# dtype = object ( Means that pandas called string as object datatype)
```

```python
runs = [1,3,65,45,76,356]
```

```python
pd.Series(runs)
```

```python
# Index is starting from 0, we can give our own index
```

```python
# Custom index

marks = [45,32,54,75,86]
subjects = ['Maths', 'Science', 'Urdu', 'computer', 'Physics']
```

```python
pd.Series(marks, index=subjects)
```

```python
pd.Series(marks, index=subjects, name='Mine marks')
```

```python
marks = pd.Series(marks, index=subjects, name='Mine marks')
```

```python
marks
```
