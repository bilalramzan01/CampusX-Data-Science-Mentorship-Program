## List in pythons

- Most Imp data type
- Data type where you can store multiple data type under one variable
- A Python list is one of the most commonly used data structures — it's an ordered, changeable (mutable) collection that can hold multiple values, of any type, all inside comma separated and in square brackets [ ].
- references array
- list is ordered
- can be nested
- item can be accessed

## List Comprehension

- short cut way to creating list

```python
l = []

for i in range(1,11):
    l.append(i)
print(l)
```

```python
l = [ i for i in range(1,11)]
```

```python
l
```

```python
v = [2,3,4]
s = -3

[s*i for i in v]
```

```python
l = [1,2,3,4,5]

[i**2 for i in l]
```

```python
[i for i in range(1,51) if i%5==0]
```

```python
### Nested if
```

```python
f1 = ['Apple', 'cherry','banana']
f2 = ['Mango', 'water melon', 'Graphes']
```

```python
[ i for i in f1 if i in f1 if i.startswith('A')]
```

```python
[[i*j for i in range(1,4)] for j in range(1,4)]
```
